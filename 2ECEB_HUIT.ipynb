## C. CATEGORY-AVERAGE VISUALIZATION

```python
track_average = df.groupby("Track")["Average"].mean()
```

```python
track_average
```

```python
gender_average = df.groupby("Gender")["Average"].mean()
```

```python
gender_average
```

```python
hometown_average = df.groupby("Hometown")["Average"].mean()
```

```python
hometown_average
```

```python
track_mean = df.groupby("Track")["Average"].mean()
track_mean
```

```python
track_mean.plot(kind="bar", figsize=[10, 6])

plt.title("Mean Average by Track")
plt.xlabel("Track")
plt.ylabel("Mean Average")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
gender_mean = df.groupby("Gender")["Average"].mean()
gender_mean
```

```python
gender_mean.plot(kind="bar", figsize=[10, 6])

plt.title("Mean Average by Gender")
plt.xlabel("Gender")
plt.ylabel("Mean Average")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
hometown_mean = df.groupby("Hometown")["Average"].mean()
hometown_mean
```

```python
hometown_mean.plot(kind="bar", figsize=[10,6])

plt.title("Mean Average by Hometown")
plt.xlabel("Hometown")
plt.ylabel("Mean Average")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## INTERPRETATION

```python
highest_track = track_mean.idxmax()
highest_gender = gender_mean.idxmax()
highest_hometown = hometown_mean.idxmax()

print("The category with the highest sample mean for Track is:", highest_track)
print("The category with the highest sample mean for Gender is:", highest_gender)
print("The category with the highest sample mean for Hometown is:", highest_hometown)
```
